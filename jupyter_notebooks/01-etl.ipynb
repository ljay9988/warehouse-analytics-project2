## Cleaning column names
Column names contain spaces, capital letters and special characters. we standardise them to make analysis easier and ensure consistency.

```python
df.columns = (
    df.columns
    .str.strip()
    .str.lower()
    .str.replace(" ", "_")
    .str.replace("(", "")
    .str.replace(")", "")
)
df.head()
```

## Convert Date columns
several date-related columns are stored as strings, we convert them to datetime objects for more accurate time-based analysis.

```python
date_cols = [
    'order_date_dateorders',
    'shipping_date_dateorders',
]

for col in date_cols:
    if col in df.columns:
        df[col] = pd.to_datetime(df[col], errors='coerce')

        df[date_cols].head()
```

## Handle missing values
now we handle the missing values in the dataset.

```python
df.loc[:, 'customer_lname'] = df['customer_lname'].fillna('Unknown')
df.loc[:, 'customer_zipcode'] = df['customer_zipcode'].fillna(0)

df.isnull().sum().sort_values(ascending=False).head(10)
```

## Feature engineering
we create new columns from existing data to support later analysis.
these engineered features help us understand shipping performance, customer behaviour, and order patterns.

Features created:
-shipping_delay: real minus scheduled shipping days
-order_month: month extracted from order date
-order_weekday: weekday name of the shipping date
-shipping_weekday: weekday name of the shipping date
-high_value_order: flag for orders above the median order_item_total
-customer_full_name: combined first and last name

```python
df['shipping_delay'] = (df['days_for_shipping_real'] - df['days_for_shipment_scheduled'])

df['order_month'] = df['order_date_dateorders'].dt.month

df['order_weekday'] = df['order_date_dateorders'].dt.day_name()

df['shipping_weekday'] = df['shipping_date_dateorders'].dt.day_name()

df['high_value_order'] = df['order_item_total'] > df['order_item_total'].median()

df['customer_full_name'] = df['customer_fname'] + ' ' + df['customer_lname']
```

## Automated ETL Pipeline
we create a ETL pipeline into a reusable function so the cleaning process can be repeated easily.

```python
def etl_pipeline():
    df = pd.read_csv("../data/raw/DataCoSupplyChainDataset.csv", encoding="ISO-8859-1", engine="python")
#clean column names
    df.columns = (
        df.columns
        .str.strip()
        .str.lower()
        .str.replace(" ", "_")
        .str.replace("(", "")
        .str.replace(")", "")
    )
#convert date columns to datetime
    date_cols = [
        'order_date_dateorders',
        'shipping_date_dateorders',
    ]
    for col in date_cols:
        df[col] = pd.to_datetime(df[col], errors='coerce')

# handle missing values
    df.loc[:, 'customer_lname'] = df['customer_lname'].fillna('Unknown')
    df.loc[:, 'customer_zipcode'] = df['customer_zipcode'].fillna(0)

# create new features
    df['shipping_delay'] = (df['days_for_shipping_real'] - df['days_for_shipment_scheduled'])
    df['order_month'] = df['order_date_dateorders'].dt.month
    df['order_weekday'] = df['order_date_dateorders'].dt.day_name()
    df['shipping_weekday'] = df['shipping_date_dateorders'].dt.day_name()
    df['high_value_order'] = df['order_item_total'] > df['order_item_total'].median()
    df['customer_full_name'] = df['customer_fname'] + ' ' + df['customer_lname']

# save cleaned data to csv
    df.to_csv("../data/cleaned/dataco_cleaned.csv", index=False)

    return df
```

```python
etl_df = etl_pipeline()
etl_df.head(10)
```
